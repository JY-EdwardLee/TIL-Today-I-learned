### 숫자 배열 회전 (D2)

```python
# 배열 만들기
n, m = map(int, input().split())
array = []
for i in range(n):
    row = list(map(int, input(f"Enter {n} numbers for row {i + 1}, separated by spaces: ").split()))
    array.append(row)

print(array)
```

```python
# 상하 좌우로 위치한 원소 합 구하기
plus_sum_H = sum(array[int(n/2)][int(n/2)-m:int(n/2)+m+1]) # n/2행 기준으로 -m~+m열까지 합
plus_sum_V = sum(list([row[column_index] for row in array[1-m:1+1+m]])) # n/2열 기준으로 -m~+m행까지 합
plus_sum = plus_sum_H + plus_sum_V - array[int(n/2)][int(n/2)]
print(plus_sum)
```
